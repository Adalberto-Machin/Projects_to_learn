```python
# quizz caculations
import numpy as np

A = np.array([[2,3],[5,6]])
np.linalg.det(A)
```

```python
np.linalg.det(A**2)
```

```python
M = np.array([[1,0,1],[0,1,0],[1,1,1]])
N = np.array([[2,8,7],[4,3,9],[1,9,5]])
np.linalg.det(M@N)
```

```python
M@N
```

```python
m = [[3,2],[5,8]]
np.cov(m, rowvar=False)
```
